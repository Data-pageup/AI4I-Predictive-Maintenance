- Clean and standardize column names.
- Verify data types.

```python
# libraries

import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/ai4i2020.csv")
```

```python
df.head(5)
```

```python
print(df.shape)
```

```python
print("missing value",df.isnull().sum().sum() )
print()
print("duplicate values", df.duplicated().sum())
```

```python
# removing the unwanted columns 

df.drop(columns = ['UDI','Product ID', 'TWF','HDF','PWF','OSF','RNF'], inplace = True)
```

```python
df.shape
```

```python
for i in df.columns:
    print(i)
```

```python
# Cleaning column names

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("[", "", regex=False)
    .str.replace("]", "", regex=False)
)
```

```python
df.columns
```

```python
df.dtypes
```

```python
# data types 
df["type"] = df["type"].astype("category")
```

```python
df.dtypes
```

## Range and Value Check conclusion 

- Air temperature ranges from 295.3 K to 304.5 K.
- Process temperature ranges from 305.7 K to 313.8 K.
- Rotational speed ranges from 1168 rpm to 2886 rpm.
- Torque ranges from 3.8 Nm to 76.6 Nm with no negative values.
- Tool wear ranges from 0 to 253 minutes.
- Machine failure contains only valid binary values (0 and 1).
- No immediately invalid or impossible values were identified.
- Some extreme values may exist, particularly in rotational speed and torque, but they should be investigated during EDA rather than removed during data cleaning.
