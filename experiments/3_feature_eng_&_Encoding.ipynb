## Feature Engineering

We created three new features by combining existing machine measurements to capture more meaningful operating conditions:

### 1. Temperature Difference (`temperature_difference_k`)

Calculated as:

Process Temperature − Air Temperature

This represents the temperature difference between the machine process and the surrounding environment, helping capture the machine's heat dissipation condition.

### 2. Power (`power_w`)

Calculated as:

Power = Torque × Angular Velocity

where:

Angular Velocity = RPM × (2π / 60)

This combines torque and rotational speed to represent the machine's overall mechanical power or operating load.

### 3. Tool Wear × Torque (`tool_wear_torque`)

Calculated as:

Tool Wear × Torque

This captures the combined effect of tool wear and mechanical load. A highly worn tool operating under high torque may experience greater mechanical stress and a higher risk of failure.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/cleaned_df.csv")
```

```python
df
```

```python
df["temperature_difference_k"] = (
    df["process_temperature_k"]
    - df["air_temperature_k"]
)
```

```python

df["power_w"] = (
    df["torque_nm"]
    * df["rotational_speed_rpm"]
    * (2 * np.pi / 60)
)
```

```python
df["tool_wear_torque"] = (
    df["tool_wear_min"]
    * df["torque_nm"]
)
```

```python
df["type"] = df["type"].map({
    "L": 0,
    "M": 1,
    "H": 2
})
```
